# Sample Code to Run Turnbull Estimator

```python
%load_ext autoreload
%autoreload 2
```

```python
import numpy as np
```

## Prepare Data

```python
lb = np.array([0, 2, 2, 4, 10])
ub = np.array([2, 6, 6, 10, np.inf])
```

## Run Turnbull Estimator

```python
import cenreg.model.nonparametric

dist = cenreg.model.nonparametric.turnbull_estimator(lb, ub)
```

## Plot Estimated Distribution

```python
%matplotlib inline

import matplotlib.pyplot as plt

x = dist.b
w = np.diff(x)
cum_p = np.append(0.0, dist.cum_p)
p = np.diff(cum_p)

fig = plt.figure()
plt.xlabel('y')
plt.ylabel('probability')
plt.xlim([dist.b[0], dist.b[-1]])
plt.ylim([0.0,1.0])
plt.bar(x[:-1], p, width=w, align='edge')
plt.show()
```
